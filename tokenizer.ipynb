```python
import os
import urllib.request

if not os.path.exists("the-verdict.txt"):
    url = ("https://raw.githubusercontent.com/rasbt/LLMs-from-scratch/refs/heads/main/ch02/01_main-chapter-code/the-verdict.txt")
    file_path = "the-verdict.txt"
    urllib.request.urlretrieve(url,file_path)
```

```python
import re
result = re.split(r'([,.:;?_!"()\']|--|\s)', open("the-verdict.txt").read())
result = [x for x in result if x.strip() != '']
len(result)
```

```python
unique = sorted(set(result))
len(unique)
```

```python
vocab = {token:integer for integer,token in enumerate(unique)}
```

```python
from SimpleTokenizerV1 import SimpleTokenizerV1
SimpleTokenizerV1 = SimpleTokenizerV1(vocab)
SimpleTokenizerV1.encode("It's the last he painted you know")
```

```python
SimpleTokenizerV1.decode([56, 2, 850, 988, 602, 533, 746, 1126, 596])
```
